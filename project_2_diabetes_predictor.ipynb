## Project 2: Predicting Diabetes

For this project we're going to use one large dataset comparing to `iris_dataset`. First we will look into the data

```python
import pandas as pd
```

```python
df = pd.read_csv('diabetes.csv')
```

```python
df.head()
```

In the diabetes table 1 means `diabetes patient` and 0 means `healthy person`. we're going to take this 2 as outputs and other 8 columns as inputs. Then we're going to predict that the person have `diabetes` or `not`.

```python
x = df[['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age']].values
```

```python
x
```

```python
y = df['Outcome'].values
```

```python
y
```

Next, we're going to convert the input and output to tensors like below:, before that we're going to import the torch dependencies

```python
import torch
import torch.nn as nn
```

```python
#converting to tensor
x = torch.tensor(x, dtype=torch.float32)
```

```python
x
```

```python
y = torch.tensor(y, dtype=torch.long)
```

```python
y
```

For this model we're going to add some more layers to train it well the 

first layer will change `8` inputs to `16` helpers

then the second layer will change `16` inputs to `8` helpers

then again to the last and third layer it will change the `8` inputs to `2` outputs

and here we're going to add one ReLU() also for better learning.

```python
class DiabeticPredictor(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer1 = nn.Linear(8, 16) #8 inputs to 16 helpers
        self.layer2 = nn.Linear(16, 8) #16 inputs to 8 helpers
        self.layer3 = nn.Linear(8, 2) # 8 inputs to 2 outputs
        self.relu = nn.ReLU()
    def forward(self, x):
        x = self.relu(self.layer1(x))
        x = self.relu(self.layer2(x))
        x = self.layer3(x)
        return x
```

```python
model = DiabeticPredictor()
```

```python
criterion = nn.CrossEntropyLoss()
```

```python
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
```

```python
for epoch in range(100000):
    y_pred = model(x)
    loss = criterion(y_pred, y)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    if (epoch+1)%5000 == 0:
        print(f"Try {epoch+1}, Mistake: {loss.item():.4f}")
```

Predicting the result

```python
test_X = x
real_y = y
guesses = model(test_X)
predictions = torch.argmax(guesses, dim=1)
total = len(x)
crct_predictions = 0
for i in range(len(x)):
    if real_y[i].item() == predictions[i].item():
        crct_predictions+=1
print(f"Total: {total} | Correct predictions: {crct_predictions}")
percentage = (crct_predictions/total)*100
print(f"Percentage: {percentage}")
```
