Building a small and simple project with a tiny dataset called `iris.csv`. It's a small and tiny dataset about flowers completely for beginners. First we'll look into the dataset and check how it looks like for loading the csv file we're going to use `pandas` module

```python
import pandas as pd
```

```python
df = pd.read_csv('Iris.csv')
```

```python
df.head()
```

You can see above the dataset contains mainly 5 columns like `SepalLengthCm`, `SepalWidthCm`, `PetalLengthCm`, `PetalWidthCm` and `Species`.

So, by using this simple dataset we'll create our simple project Flower Guesser. For inputs we will use:

input: 2 measurements like petal length and petal width

output: by using this inputs we'll guess the flower species

So first we will check for how many species there in dataset

```python
df.Species.unique()
```

So, there is three species so our goal is to create a flower guesser model to guess the flower species based on the `petal width` and `petal height` so, let's start the process.

First we'll install the torch

```python
!pip install torch
```

```python
#importing necessary libararies
import torch
import torch.nn as nn
```

After this we need to assign the input and output values by using dataset like below

```python
x = df[['PetalLengthCm', 'PetalWidthCm']].values
```

```python
x
```

Now we taken the `x` as our input next we need to taken our species as our output before that we're going to create a new column to map 0, 1, 2 to our species 'Iris-setosa', 'Iris-versicolor', 'Iris-virginica'. For, that we're going to use the below code:

```python
#mapping species names to num
names_to_num = {"Iris-setosa":0, "Iris-versicolor":1, "Iris-virginica":2}

#creating new column and adding the data based on species to the dataframe using pandas
df['species_to_num'] = df['Species'].map(names_to_num)
```

```python
df.head()
```

so, the mapping done next we will assign this values to our output variable `y`

```python
y = df['species_to_num'].values
```

```python
y
```

Now, we taken our inputs and outputs next we're going to convert this to tensors like below

```python
x = torch.tensor(x, dtype=torch.float32)
```

x

```python
y = torch.tensor(y, dtype=torch.long)
```

```python
y
```

ok now we have inputs and outputs next we will build a model like below

```python
# building the model
class FlowerGuesser(nn.Module):
    def __init__(self):
        super(FlowerGuesser, self).__init__()
        self.layer = nn.Linear(2, 3) #2 inputs petal width and height and 3 outputs 1, 2, 3

    def forward(self, x):
        x = self.layer(x)
        return x
```

```python
model = FlowerGuesser()
```

now we created a model next we need to train it

```python
#training the model
criterion = nn.CrossEntropyLoss()
```

```python
optimizer = torch.optim.SGD(model.parameters(), lr=0.1)
```

```python
for epoch in range(200):
    y_pred = model(x)
    loss = criterion(y_pred, y)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    if (epoch+1)%50 == 0:
        print(f"Try {epoch+1}, Mistake: {loss.item():.4f}")
```

Training completed next we're going to save the model

```python
torch.save(model.state_dict(), "flower_guesser.pth")
```

Now we successfully save the model, next we will load the model and use it on our program

```python
#loading the saved model
trained_model = FlowerGuesser()
trained_model.load_state_dict(torch.load("flower_guesser.pth"))
```

```python
trained_model.eval()
```

So, finally we trained the model next we're going to test with some samples

```python
df.head()
```

```python
test_x = torch.tensor([[5.1, 2.3]], dtype=torch.float32)
```

```python
guesses = trained_model(test_x)
```

```python
prediction = torch.argmax(guesses, dim=1)
```

```python
print(prediction.item())
```
