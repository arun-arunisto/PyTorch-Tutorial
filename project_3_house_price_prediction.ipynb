## Normalizing or Standardizing the Data

Without normalizing, the model thinks `Area` is way more important because its numbers are so big. like 

**Before Normalizing**

- `Area=2000` vs. `Bedrooms=3` the model might only care about `Area` because `2000` is so much bigger than `3`.
- It could miss that `3` bedrooms vs. `2` bedrooms matters too!

**After Normalizing**

- `Area` might go from `[2000, 1500]` to `[0.5, -0.5]` (average = 0, spread = 1)
- `Bedrooms` might go from `[3, 2]` to `[0.7, -0.7]`.
- Now they’re on the same scale `0.5` vs. `0.7` so the model listens to all clues fairly!

**Note:** for `YearBuilt` also same.

```python
x = (x-x.mean(axis=0))/x.std(axis=0)
```

```python
x
```

```python
y_mean = y.mean()
y_std = y.std()
```

```python
y = (y-y_mean)/y_std
```

Now we normalized data next, we need to convert the input and output to tensors like below:

```python
import torch
import torch.nn as nn
```

```python
x = torch.tensor(x, dtype=torch.float32)
y = torch.tensor(y, dtype=torch.float32).reshape(-1, 1)
```

```python
x
```

```python
y
```

Next we're going to create a model

```python
class HousePricePredictor(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer1 = nn.Linear(8, 64)
        self.layer2 = nn.Linear(64, 64)
        self.layer3 = nn.Linear(64, 32)
        self.layer4 = nn.Linear(32, 32)
        self.layer5 = nn.Linear(32, 16)
        self.layer6 = nn.Linear(16, 16)
        self.layer7 = nn.Linear(16, 8)
        self.layer8 = nn.Linear(8, 1)
        self.relu = nn.ReLU()
    def forward(self, x):
        x = self.relu(self.layer1(x))
        x = self.relu(self.layer2(x))
        x = self.relu(self.layer3(x))
        x = self.relu(self.layer4(x))
        x = self.relu(self.layer5(x))
        x = self.relu(self.layer6(x))
        x = self.relu(self.layer7(x))
        x = self.layer8(x)
        return x
```

```python
model = HousePricePredictor()
```

```python
criterion = nn.MSELoss()
```

```python
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
```

Now let's start training

```python
for epoch in range(200000):
    y_pred = model(x)
    loss = criterion(y_pred, y)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    if (epoch+1)%5000 == 0:
        print(f"Try {epoch+1} Mistake: {loss.item():.4f}")
        
```

```python
with torch.no_grad():
    y_pred = model(x[:5])
    y_pred_dollars = y_pred * y_std + y_mean  # Back to dollars
    y_dollars = y[:5] * y_std + y_mean
    print("Guessed Prices:", y_pred_dollars.flatten().tolist())
    print("Real Prices:", y_dollars.flatten().tolist())
```
