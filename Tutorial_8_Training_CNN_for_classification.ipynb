## Dataset Pre-processing

Here, we normalize the image data to the range [0,1] with `mean` and `std` of our Dataset. This is very common when working with image data which helps the model train more efficiently.

````python
mean = [0.4368, 0.4336, 0.3294]  #mean and std of this Monkey Species dataset
std = [0.2457, 0.2413, 0.2447]
````

In PyTorch, when you see `mean` and standard deviation (`std`) values like [0.4368, 0.4336, 0.3294] and [0.2457, 0.2413, 0.2447] for a custom dataset (e.g., a Monkey Species dataset), these are not random nor are they default values. They are typically calculated specifically for that dataset.

In deep learning, when working with image datasets, it’s common to normalize the pixel values of the images before feeding them into a neural network. Normalization (often subtracting the mean and dividing by the standard deviation) helps the model train faster and more stably by ensuring the input data has a mean around 0 and a standard deviation around 1. For RGB images, this is done independently for each channel (Red, Green, Blue), which is why you see three values for both mean and std.

**How these values are calculated**

1. Load the Dataset: All images in the dataset (e.g., the Monkey Species dataset) are loaded.
2. Compute Per-Channel Statistics:
  - For each channel (R, G, B), calculate the average pixel intensity across all images (this gives the mean).
  - Then, calculate the standard deviation of the pixel intensities for each channel across all images.
3. Result: You end up with a mean and std vector, each with three values corresponding to the R, G, and B channels.

```Python
def get_mean_std(train_loader, img_size=(224, 224), num_workers=2):

    batch_mean = torch.zeros(3)
    batch_mean_sqrd = torch.zeros(3)

    for batch_data, _ in train_loader:
        batch_mean += batch_data.mean(dim=(0, 2, 3))  # E[batch_i]
        batch_mean_sqrd += (batch_data**2).mean(dim=(0, 2, 3))  #  E[batch_i**2]

    mean = batch_mean / len(train_loader)

    var = (batch_mean_sqrd / len(train_loader)) - (mean**2)

    std = var**0.5
    print("mean: {}, std: {}".format(mean, std))

    return mean, std
```

## CNN model implementation in PyTorch

Let’s start with the very first convolutional layer in the first convolutional block. To define a convolutional layer in PyTorch, we call the `nn.Conv2D()` function, which accepts several input arguments. First, we define the layer to have 32 filters. The kernel size for first filter is 5 and the subsequent layers filter is 3 (which is interpreted as 3x3). We can use a padding option called same, which will pad the input tensor so that the output of the convolution operation has the same spatial size as the input. This is not required, but it’s commonly used. if you don’t explicitly specify this padding option, then the default behavior has no padding, and therefore, the spatial size of output from the convolutional layer will be slightly smaller than the input size. After each convolutional layer, we add a `BatchNorm2d` layer, which normalizes the activations of the previous layer at each batch, thereby improving the training speed and stability of the network. We use a `ReLU` activation function in all the layers in the Network except for the output layer. This sequence of `Conv2d` followed by `BatchNorm2d` and `ReLU` is called `Conv2dNormActivation` and torchvision has a convenience function to implement this by `torchvision.ops.Conv2dNormActivation()`.

There is also another alternative approach to specify Conv2d layers with `nn.LazyConv2d()` where the input_channels is automatically inferred from the previous `Conv2d` layers output_channels.

The first two convolutional layers has 32 filters each, and then we follow that with a max pooling layer that has a window size of (2x2), so the output shape from this first convolution block is (218 x 218 x 32). Next, we have the second convolutional block, has 64 and 128 filters in each convolutional layer instead of 32, and then finally, the third and fourth convolutional block has 256 and 512 filters.

**Note:** The number of filters in each convolutional layer is something that you will need to experiment with. A larger number of filters allows the model to have a greater learning capacity, but this also needs to be balanced with the amount of data available to train the model. Adding too many filters (or layers) can lead to overfitting, one of the most common issues encountered when training models.

The final layer in the feature extractor is the `nn.AdaptiveAvgPool2d()`, which applies a 2D adaptive average pooling over an input image composed of several input channels. This layer ensures that the output has a fixed size of H x W, regardless of the input size. The number of output features is equal to the number of input channels. This is particularly useful for making the subsequent fully connected layers agnostic to the input size.

Before we define the fully connected layers for the classifier, we need to first flatten the two-dimensional activation maps that are produced by the last convolutional layer (which have a spatial shape of 3x3 with 512 channels). This is accomplished by calling the `nn.Flatten()` function to create a 1-dimensional vector of length 4608. We then add a densely connected layer with 256 neurons and a fully connected output layer with 10 neurons because we have ten classes in our dataset.

We will show the different ways in which a `Conv2d -> BatchNorm -> ReLU`

````python
class MyModel(nn.Module):
  """
  This defines a custom model inheriting from nn.Module, the base class for all neural networks in PyTorch.
  """
  def __init__(self):
    super().__init__()
    """
    super().__init__() initializes the parent class, ensuring the model can leverage PyTorch’s built-in functionality (e.g., parameter tracking).
    """
    """
    Model Architecture:
      - Convolutional Layers: Feature extraction using convolutions, normalization, activation, pooling, and dropout.
      - Pooling and Flattening: Transition from spatial features to a flat vector.
      - Fully Connected Layers: Classification head to produce the final output.
    """
    self._model = nn.Sequential(

        #---------------------- Convolution Layers ----------------------
        #The convolutional part is split into three blocks, each increasing the number of filters (feature maps) and reducing the spatial dimensions via pooling.

        #-----------------------------------------------
        # Conv2d Norm Activation Block1: 32 Filters, MaxPool.
        #-----------------------------------------------
        nn.Conv2d(in_channels = 3, out_channels = 32, kernel_size = 5),
        # First nn.Conv2d: Takes an input with 3 channels (e.g., RGB image), applies 32 filters of size 5x5, producing 32 feature maps.
        # The spatial size decreases slightly due to no padding (default).
        nn.BatchNorm2d(32),
        # nn.BatchNorm2d(32): Normalizes the 32 feature maps across the batch, stabilizing training.
        nn.ReLU(inplace = True),
        # nn.ReLU(inplace=True): Applies the ReLU activation function (max(0, x)) to introduce nonlinearity; inplace=True modifies the tensor in-place to save memory.
        nn.Conv2d(in_channels = 32, out_channels = 32, kernel_size = 3),
        # Second nn.Conv2d: Takes the 32 channels, applies 32 new 3x3 filters, keeping the number of channels at 32 but refining the features.
        nn.BatchNorm2d(32),
        nn.ReLU(inplace = True),
        # Another BatchNorm2d and ReLU pair.
        nn.MaxPool2d(kernel_size = 2),
        # nn.MaxPool2d(kernel_size=2): Reduces the spatial dimensions (height and width) by half, e.g., from 32x32 to 16x16, by taking the maximum value in each 2x2 region.
        #-------------------------------------------------------------------
        # Conv2d Norm Activation Block 2: 64,128 Filters, MaxPool, Dropout(p=0.25)
        #-------------------------------------------------------------------
        nn.LazyConv2d(out_channels = 64, kernel_size = 3),
        # nn.LazyConv2d(out_channels=64, kernel_size=3): A "lazy" convolution layer that infers the input channels automatically (32 from the previous layer) and outputs 64 channels with 3x3 filters.
        # This is useful when chaining layers without manually specifying in_channels.
        nn.BatchNorm2d(64),
        nn.ReLU(inplace = True),
        # Followed by BatchNorm2d and ReLU.
        nn.LazyConv2d(out_channels = 128, kernel_size = 3),
        # nn.LazyConv2d(out_channels=128, kernel_size=3): Increases the channels to 128, further extracting more complex features.
        nn.BatchNorm2d(128),
        nn.ReLU(inplace = True),
        # Another BatchNorm2d and ReLU.
        nn.MaxPool2d(kernel_size = 2),
        # nn.MaxPool2d(kernel_size=2): Halves the spatial dimensions again (e.g., 16x16 to 8x8).
        #--------------------------------------------------
        # Conv2d Norm Activation Block 3: 128,256,512 Filters, MaxPool.
        #--------------------------------------------------
        Conv2dNormActivation(in_channels = 128, out_channels=256, kernel_size = 3),
        Conv2dNormActivation(in_channels = 256, out_channels=256, kernel_size = 3),
        nn.MaxPool2d(kernel_size = 2),
        Conv2dNormActivation(in_channels = 256, out_channels=512, kernel_size = 3),
        nn.MaxPool2d(kernel_size = 2),
        # Conv2dNormActivation: This is a PyTorch utility (from torchvision.ops) that combines Conv2d, BatchNorm2d, and ReLU into one layer for convenience.
        # First instance: Takes 128 input channels (from Block 2), outputs 256 channels with 3x3 filters.
        # Second instance: Keeps 256 channels, refining features further.
        # nn.MaxPool2d(kernel_size=2): Reduces spatial size (e.g., 8x8 to 4x4).
        # Third Conv2dNormActivation: Increases to 512 channels, capturing highly abstract features.
        # Another MaxPool2d: Reduces spatial size further (e.g., 4x4 to 2x2).
        #---------------------- Feed Forward Layers --------------------
        nn.AdaptiveAvgPool2d(output_size=(3,3)),
        # Reduces the spatial dimensions of the feature maps to a fixed 3x3 size, regardless of the input size after convolutions.
        # Output shape becomes (batch_size, 512, 3, 3) (512 channels from the last conv layer).
        #------------------------------------
        # Flatten the convolutional features.
        #------------------------------------
        nn.Flatten(),
        # Collapses the 512x3x3 tensor into a 1D vector per sample.
        # Output shape: (batch_size, 512*3*3) = (batch_size, 4608).
        #--------------------
        # Classification Head.
        #--------------------
        nn.Linear(in_features = 512*3*3, out_features = 256),
        # Takes the flattened 4608 features (512 * 3 * 3) and reduces them to 256, acting as a bottleneck to compress information.
        # No activation function here (linear transformation only).
        nn.Linear(in_features = 256, out_features = 10)
        # Maps the 256 features to 10 outputs, corresponding to 10 classes (e.g., 10 monkey species).
        # The raw outputs (logits) are typically passed through a softmax or cross-entropy loss during training for classification.
    )
  """
  The forward method defines how input x (e.g., a batch of images with shape (batch_size, 3, height, width)) flows through the model.
  Since _model is an nn.Sequential, it automatically passes x through all layers in order, returning the final output with shape (batch_size, 10).
  """
  def forward(self,x):
      return self._model(x)
````

## Overall Architecture Summary
- **Input:** An RGB image (3 channels).
- **Convolutional Blocks:** Extracts features with increasing complexity (32 → 64 → 128 → 256 → 512 filters), reducing spatial size via pooling.
- **Pooling/Flattening:** Fixes spatial size to 3x3, then flattens to a 4608-dimensional vector.
- **Fully Connected Layers:** Compresses features (4608 → 256) and outputs 10 class scores.
- **Output:** 10 logits for classification.

This is a classic CNN design, similar to VGG-style architectures, with batch normalization for stability and ReLU for nonlinearity.

## Model Training and Evaluation

````python
def train(model, train_loader):
    model.train()
    model.to(DEVICE)

    running_loss = 0
    correct_predictions = 0
    total_train_samples = 0

    for images, labels in tqdm(train_loader, desc="Training"):
        images, labels = images.to(DEVICE), labels.to(DEVICE)
        optimizer.zero_grad()
        outputs = model(images)
        loss = F.cross_entropy(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item()
        _, predicted = torch.max(outputs.data, dim=1)
        total_train_samples += labels.shape[0]
        correct_predictions += (predicted == labels).sum().item()

    train_avg_loss = running_loss / len(train_loader)
    train_accuracy = 100 * correct_predictions / total_train_samples
    return train_avg_loss, train_accuracy
````

````python
def validation(model, val_loader):
    model.eval()
    model.to(DEVICE)

    running_loss = 0
    correct_predictions = 0
    total_val_samples = 0

    for images, labels in tqdm(val_loader, desc="Validation"):
        images, labels = images.to(DEVICE), labels.to(DEVICE)

        with torch.no_grad():
             outputs = model(images)

        loss = F.cross_entropy(outputs, labels)
        running_loss += loss.item()
        _, predicted = torch.max(outputs.data, dim=1)
        total_val_samples += labels.shape[0]
        correct_predictions += (predicted == labels).sum().item()

    val_avg_loss = running_loss / len(val_loader)
    val_accuracy = 100 * correct_predictions / total_val_samples
    return val_avg_loss, val_accuracy
````

````python
def main(model, train_loader, val_loader):

    train_losses, val_losses = [], []
    train_accuracies, val_accuracies = [], []

    best_val_acc = 0.0
    best_weights = None

    for epoch in range(train_config.num_epochs):
        train_loss, train_accuracy = train(model, train_loader)
        val_loss, val_accuracy = validation(model, val_loader)


        train_losses.append(train_loss)
        train_accuracies.append(train_accuracy)
        val_losses.append(val_loss)
        val_accuracies.append(val_accuracy)

        print(f"Epoch {epoch+1:0>2}/{train_config.num_epochs} - Train Loss: {train_loss:.4f}, Train Accuracy: {train_accuracy:.2f}% - Val Loss: {val_loss:.4f}, Val Accuracy: {val_accuracy:.2f}%")

        # Logging metrics to tensorboard
        writer.add_scalar('Loss/train', train_loss)
        writer.add_scalar('Loss/val', val_loss)
        writer.add_scalar('Accuracy/train', train_accuracy)
        writer.add_scalar('Accuracy/val', val_accuracy)

        if val_accuracy > best_val_acc:
            best_val_acc = val_accuracy
            best_weights =  model.state_dict()
            print(f"Saving best model...💾")
            torch.save(best_weights, "best.pt")

    return train_losses, train_accuracies, val_losses, val_accuracies
````

## Inference

````python
def prediction(model, val_loader):

    model.eval()
    model.to(DEVICE)

    all_images, all_labels = [], []
    all_pred_indices, all_pred_probs = [], []

    for images, labels in val_loader:
        images, labels = images.to(DEVICE), labels.to(DEVICE)

        with torch.inference_mode():
             outputs = model(images)

        prob = F.softmax(outputs,dim=1)
        pred_indices = prob.data.max(dim=1)[1]
        pred_probs = prob.data.max(dim=1)[0]

        all_images.append(images.cpu())
        all_labels.append(labels.cpu())
        all_pred_indices.append(pred_indices.cpu())
        all_pred_probs.append(pred_probs.cpu())


    return (torch.cat(all_images).numpy(),
            torch.cat(all_labels).numpy(),
            torch.cat(all_pred_indices).numpy(),
            torch.cat(all_pred_probs).numpy())
````

We will need to denormalize our pixel values values by multiplying `std` and adding `mean` value to our image for `Matplotlib` visualization. `np.clip` makes the pixel values to be in range of [0,1]

````python
def denormalize(image):
    mean_ar = np.array(mean)
    std_ar = np.array(std)
    image = image * std_ar + mean_ar
    return np.clip(image, 0,1)
````

````python
def visualise_predictions(sample_images,sample_gt_labels, pred_indices, pred_probs, num_images =5):

    fig = plt.figure(figsize = (20,5))

    for i in range(num_images):
        idx = random.randint(0, len(sample_images) -1)
        image = sample_images[idx].transpose(1,2,0) #(C,H,W) --> (H,W,C)
        label = sample_gt_labels[idx]
        pred_idx = pred_indices[idx]
        pred_prob = pred_probs[idx]

        image = denormalize(image)

        ax = fig.add_subplot(1, num_images, i+1)
        ax.imshow(image)
        ax.set_title(f"GT: {class_mapping[label]}\nPred: {class_mapping[pred_idx]} ({pred_prob:.2f})")
        ax.axis('off')

    plt.show()
````
